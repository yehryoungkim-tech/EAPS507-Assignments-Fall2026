# Madhava-Leibniz π calculation


This notebook is designed to be a live demonstration of Python fundamentals applied to calculating π using the [Madhava-Leibniz series](https://en.wikipedia.org/wiki/Approximations_of_%CF%80#Middle_Ages)

The approximation is simple:

$$ \pi \approx 4 \cdot \left(1 - \frac{1}{3} + \frac{1}{5} - \frac{1}{7} + \frac{1}{9} \ldots -(-1)^n \frac{1}{2n - 1}  \right) $$

## Part 1 — Numerical Calculation and Error Analysis

```python
import math
import matplotlib.pyplot as plt

N = 5000

iterations = []
pi_values = []
errors = []

pi_sum = 0

for n in range(1, N + 1):

    if n == 1:
        term = 1
    else:
        term = -((-1)**n) / (2*n - 1)

    pi_sum = pi_sum + term

    calculated_pi = 4 * pi_sum
    absolute_error = abs(calculated_pi - math.pi)

    iterations.append(n)
    pi_values.append(calculated_pi)
    errors.append(absolute_error)

print("Calculated pi:", pi_values[-1])
print("Expected pi:", math.pi)
print("Absolute error:", errors[-1])
```

```python
plt.figure(figsize=(8, 5))

plt.plot(iterations, errors)

plt.xscale("log")
plt.yscale("log")

plt.xlabel("Number of Iterations")
plt.ylabel("Absolute Error")
plt.title("Error of Madhava-Leibniz Pi Approximation")

plt.grid(True)

plt.show()
```

```python
accuracy_threshold = 0.001

required_iteration = None

for n, error in zip(iterations, errors):

    if error < accuracy_threshold:
        required_iteration = n
        break

print("Accuracy threshold:", accuracy_threshold)
print("Iterations required:", required_iteration)
```

### Reasonable Accuracy

I defined reasonable accuracy as an absolute error less than 0.001. I chose this value because being within one-thousandth of the expected value is accurate enough to clearly show that the approximation is approaching pi. The calculation required approximately 1000 iterations to reach this level.

## Part 2 — Function


```python
def calculate_pi(n_iterations):
  """
  Calculate an approximation of pi using the Madhava-Leibniz series.
  """

  pi_sum = 0

  for n in range(1, n_iterations + 1):

      if n == 1:
          term = 1
      else:
          term = -((-1)**n) / (2*n - 1)

      pi_sum = pi_sum + term

  return 4 * pi_sum
```

```python
print("10 iterations:", calculate_pi(10))
print("100 iterations:", calculate_pi(100))
print("1000 iterations:", calculate_pi(1000))
```
